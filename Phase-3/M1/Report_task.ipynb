## Task 1:  What generative AI is, what an AI agent is, and one example

**Generative AI (gen AI):** Generates content such as text, audio, images, and video using foundation models (LLMs) to extract insights.

**AI Agent:** A software component with agency to act on behalf of a user or system to perform tasks, orchestrate complex workflows, coordinate activities among multiple agents, and evaluate answers.

*Example:* In customer service, a gen AI agent uses customer data, call center transcripts, and company policies to assist call center staff in responding to customer questions.

## Task 2: Agent architecture and decision-making

An AI agent starts with a user goal, plans how to achieve it, and breaks complex work into smaller subtasks. A manager agent can assign subtasks to specialized agents, which coordinate using organizational or external data to complete assignments.

During decision-making, agents use reasoning and available tools (APIs, web browsers) to gather information and perform actions. They also use learning and reflection to review results and improve performance, incorporating user feedback for iterative refinement.


## Task 3:  Compare different agent frameworks and design patterns

### Frameworks:

* **LangChain:** Facilitates LLM tool integration, supporting chatbots, document analysis, modular workflows, and prompt management.

* **LlamaIndex:** Open-source framework for building LLM agents over enterprise data, supporting 160+ formats and RAG capabilities.

* **AutoGen:** Developed by Microsoft for multi-agent workflows, providing a Core API for message passing and AgentChat API for group chats.


### Design Patterns:

1. **Reflection:** The AI reviews its own work to find mistakes and improve it, helping reduce errors and improve reliability.

2. **Tool Use:** The AI uses external tools, such as web search or APIs, to get information or perform tasks it cannot do on its own.

3. **ReAct:** **Reason and Act** combines step-by-step reasoning with actions. The AI reasons about what to do and then takes actions, such as using tools.

4. **Planning:** The AI breaks down big tasks into smaller steps and creates a plan to tackle them.

5. **Multi-Agent Collaboration:** Several AI agents work together, with each agent having a specific role, such as planning, coding, or checking the work.
