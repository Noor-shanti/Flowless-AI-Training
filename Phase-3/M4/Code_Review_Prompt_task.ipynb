# Task: Build a Code Review Prompt to Use in an Agent
#### Phase 3 - Module 4: Prompt Engineering & Context Engineering

## 1. Define the Agent Role

### Agent Role
**Code Review Agent** - an AI software engineering assistant that reviews code and gives useful feedback.

The agent should:
- Find bugs and correctness issues.
- Check readability and maintainability.
- Suggest practical improvements.
- Explain why each issue matters.
- Mention performance or security concerns when relevant.
- Avoid inventing issues that are not supported by the provided code and context.

## 2. Select Code Snippets for Review

For this task, I selected a small Python function that calculates the average of a list of numbers.

This snippet is suitable because it is simple, but it still has realistic issues that an agent can detect, such as:
- handling empty input
- improving readability
- simplifying the loop

```python
code_snippet = '''
def get_average(numbers):
    total = 0
    for i in range(len(numbers)):
        total += numbers[i]
    return total / len(numbers)
'''

print(code_snippet)
```

## 3. Write a Base Prompt

The base prompt starts simple. It gives the agent a role, explains the task, and separates the code from the instructions using delimiters like `<CODE>` tags.

```python
base_prompt = '''
You are a Code Review Agent.

Review the code below inside the <CODE> tags.
Find important issues, explain why they matter, and suggest improvements.
Focus on correctness, readability, maintainability, performance, and security.

<CODE>
{code}
</CODE>
'''

print(base_prompt.format(code=code_snippet))
```

## 4. Enhance with Context Engineering

In this step, I improve the base prompt by adding relevant context.

The agent should not only see the code. It should also know:
- the programming language
- the goal of the code
- the expected behavior
- the coding style
- the review scope

This helps the agent give a more accurate review without adding unrelated information.

```python
context = {
    "language": "Python 3",
    "project_goal": "Calculate the average of a list of numbers.",
    "expected_behavior": "Return a numeric average and handle empty input clearly.",
    "style": "Readable and maintainable code for a small project.",
    "review_scope": [
        "correctness",
        "readability",
        "maintainability",
        "performance",
        "security"
    ]
}

context
```

```python
enhanced_prompt = '''
You are a Code Review Agent.

## Task
Review the provided code and:
1. Find important issues.
2. Explain why each issue matters.
3. Suggest a simple improvement.
4. Provide an improved version of the code if needed.

## Review Criteria
- Correctness
- Readability
- Maintainability
- Performance
- Obvious security concerns

Do not invent issues.
Keep the same intended behavior unless a change is needed to fix a problem.

## Project Context
<CONTEXT>
Language: {language}
Project goal: {project_goal}
Expected behavior: {expected_behavior}
Style: {style}
Review scope: {review_scope}
</CONTEXT>

## Code To Review
<CODE>
{code}
</CODE>

## Output Format
Return a JSON object with these keys:
- summary
- issues: a list of issues with category, severity, explanation, and suggestion
- improved_code

Use only the information available in the context and code.
'''

formatted_enhanced_prompt = enhanced_prompt.format(
    language=context["language"],
    project_goal=context["project_goal"],
    expected_behavior=context["expected_behavior"],
    style=context["style"],
    review_scope=", ".join(context["review_scope"]),
    code=code_snippet
)

print(formatted_enhanced_prompt)
```

### What changed from the base prompt?

The enhanced prompt adds useful context and asks for JSON output. It also gives clearer instructions and tells the agent not to invent issues.

## 5. Experiment with Prompt Variations

In this step, I test different versions of the prompt. Each version keeps the same code review task, but changes the style or output format.

### Variation A - Concise Review
This version is short and focuses only on the most important issues.

```python
prompt_variation_a = '''
You are a Code Review Agent.
Review the code inside <CODE> delimiters.
List only the most important issues.
For each issue, give the category, severity, explanation, and suggested fix.
Be concise and do not invent issues.

<CODE>
{code}
</CODE>
'''.format(code=code_snippet)

print(prompt_variation_a)
```

### Variation B - Teaching Review
This version explains the issues in a simple way for a learner.

```python
prompt_variation_b = '''
You are a Code Review Agent acting as a patient programming mentor.

Review the code below. For every meaningful issue:
1. Name the issue.
2. Explain what is happening in simple technical language.
3. Explain why it matters.
4. Show how to improve it.
5. Give a short learning takeaway.

Do not criticize code without explaining the reason, and do not invent issues.

<CODE>
{code}
</CODE>
'''.format(code=code_snippet)

print(prompt_variation_b)
```

### Variation C - Structured Agent Review
This version asks for JSON output, so it is easier for an agent to use.

```python
prompt_variation_c = '''
You are a Code Review Agent.

Analyze the code using the supplied project context.
Prioritize correctness first, then readability, maintainability, performance, and obvious security risks.
Only report issues supported by the provided information.

Return valid JSON with exactly these top-level keys:
{{
  "summary": "...",
  "issues": [
    {{
      "category": "...",
      "severity": "low|medium|high",
      "explanation": "...",
      "suggestion": "..."
    }}
  ],
  "improved_code": "..."
}}

<CONTEXT>
Language: {language}
Goal: {goal}
Expected behavior: {expected}
</CONTEXT>

<CODE>
{code}
</CODE>
'''.format(
    language=context["language"],
    goal=context["project_goal"],
    expected=context["expected_behavior"],
    code=code_snippet
)

print(prompt_variation_c)
```

### Comparing the Variations

- **Variation A:** best for a quick review.
- **Variation B:** best for explaining issues to a learner.
- **Variation C:** best for agent use because the output is structured.

For this task, **Variation C** is the most suitable variation.
